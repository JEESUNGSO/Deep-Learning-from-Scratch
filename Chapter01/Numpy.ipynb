```python
import numpy as np

x = np.array([1.5, 2.2, 5.5, 2.8])
y = np.array([1.0, 2.0, 6.1, 2.2])

print(x)        # [1.5 2.2 5.5 2.8]
print(type(x))  # <class 'numpy.ndarray'>
print(x + y)    # [ 2.5  4.2 11.6  5. ]
print(x - y)    # [ 0.5  0.2 -0.6  0.6]
print(x * y)    # [ 1.5   4.4  33.55  6.16]
print(x / y)    # [1.5        1.1        0.90163934 1.27272727]
print(x / 2.0)  # [0.75 1.1  2.75 1.4 ]

A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [10, 20],
    [30, 40]
])

print(A)
# [[1 2]
#  [3 4]]
print(A.shape)  # (2, 2)
print(A.dtype)  # int64
print(A + B)
# [[11 22]
#  [33 44]]
print(A * B)
# [[ 10  40]
#  [ 90 160]]
print(A * 0.5) # Broadcast: Make same dimension as A with lower dimension Vector or scalar.
# [[0.5 1. ]
#  [1.5 2. ]]

k = np.array([2, 4])

print(A * k) # Broadcast
# [[ 2  8]
#  [ 6 16]]

print(A.flatten())  # [1 2 3 4]


A = A.flatten()

print(A > 2.5) # [False False  True  True]
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 6, 0.1)
y = np.sin(x)

plt.plot(x,y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 6, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin")
plt.plot(x, y2, linestyle="--", label="cos")
plt.xlabel("x")
plt.ylabel("y")
plt.title("sin & cos")
plt.legend()
plt.show()
```
